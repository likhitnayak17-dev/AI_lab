```python


def alpha_beta(depth, node, is_max, values, alpha, beta, max_depth):

    # Base condition: leaf node
    if depth == max_depth:
        return values[node]

    if is_max:
        best = float('-inf')

        for i in range(2):
            val = alpha_beta(depth + 1, node * 2 + i,
                             False, values, alpha, beta, max_depth)

            best = max(best, val)
            alpha = max(alpha, best)

            # Alpha-Beta Pruning
            if beta <= alpha:
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            val = alpha_beta(depth + 1, node * 2 + i,
                             True, values, alpha, beta, max_depth)

            best = min(best, val)
            beta = min(beta, best)

            # Alpha-Beta Pruning
            if beta <= alpha:
                break

        return best


# Taking input from user
max_depth = int(input("Enter the depth of the tree: "))

n = 2 ** max_depth

values = []

print("Enter the leaf node values:")
for i in range(n):
    values.append(int(input(f"Node {i+1}: ")))

# Apply Alpha-Beta Pruning
result = alpha_beta(0, 0, True, values,
                    float('-inf'), float('inf'), max_depth)

print("\nOptimal value:", result)
```
