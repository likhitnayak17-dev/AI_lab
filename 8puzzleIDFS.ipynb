```python
# 8 Puzzle Problem using Depth First Search (DFS)

def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


# Generate possible moves
def get_moves(state):
    moves = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    # Move blank UP
    if row > 0:
        new_state = state[:]
        new_state[blank], new_state[blank - 3] = \
            new_state[blank - 3], new_state[blank]
        moves.append(new_state)

    # Move blank DOWN
    if row < 2:
        new_state = state[:]
        new_state[blank], new_state[blank + 3] = \
            new_state[blank + 3], new_state[blank]
        moves.append(new_state)

    # Move blank LEFT
    if col > 0:
        new_state = state[:]
        new_state[blank], new_state[blank - 1] = \
            new_state[blank - 1], new_state[blank]
        moves.append(new_state)

    # Move blank RIGHT
    if col < 2:
        new_state = state[:]
        new_state[blank], new_state[blank + 1] = \
            new_state[blank + 1], new_state[blank]
        moves.append(new_state)

    return moves


# DFS function
def dfs(start, goal):
    stack = [(start, [])]
    visited = set()

    while stack:

        state, path = stack.pop()

        state_tuple = tuple(state)

        if state_tuple in visited:
            continue

        visited.add(state_tuple)

        # Check goal
        if state == goal:
            return path + [state]

        # Generate next states
        for next_state in get_moves(state):
            if tuple(next_state) not in visited:
                stack.append((next_state, path + [state]))

    return None


# Initial state
start = [5, 4, 6,
         1, 8, 7,
         3, 2, 0]

# Goal state
goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]


# Solve the puzzle
solution = dfs(start, goal)


# Display solution
if solution:
    print("Solution found!")
    print("Number of steps:", len(solution) - 1)
    print()

    for i, state in enumerate(solution):
        print("Step", i)
        print_puzzle(state)

else:
    print("No solution found.")
```
