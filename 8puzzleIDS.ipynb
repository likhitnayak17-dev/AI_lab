```python
# 8-Puzzle Problem using Iterative Deepening Search (IDS)

def print_puzzle(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


# Generate possible moves
def get_moves(state):
    moves = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    # Move blank UP
    if row > 0:
        new_state = state[:]
        new_state[blank], new_state[blank - 3] = \
            new_state[blank - 3], new_state[blank]
        moves.append(new_state)

    # Move blank DOWN
    if row < 2:
        new_state = state[:]
        new_state[blank], new_state[blank + 3] = \
            new_state[blank + 3], new_state[blank]
        moves.append(new_state)

    # Move blank LEFT
    if col > 0:
        new_state = state[:]
        new_state[blank], new_state[blank - 1] = \
            new_state[blank - 1], new_state[blank]
        moves.append(new_state)

    # Move blank RIGHT
    if col < 2:
        new_state = state[:]
        new_state[blank], new_state[blank + 1] = \
            new_state[blank + 1], new_state[blank]
        moves.append(new_state)

    return moves


# Depth Limited Search
def depth_limited_search(state, goal, depth, path):

    # Goal test
    if state == goal:
        return path

    # Depth limit reached
    if depth == 0:
        return None

    # Explore next states
    for next_state in get_moves(state):

        if next_state not in path:
            result = depth_limited_search(
                next_state,
                goal,
                depth - 1,
                path + [next_state]
            )

            if result is not None:
                return result

    return None


# Iterative Deepening Search
def ids(start, goal):

    depth = 0

    while True:

        print("Searching at depth:", depth)

        result = depth_limited_search(
            start,
            goal,
            depth,
            [start]
        )

        if result is not None:
            return result

        depth += 1


# Initial state
start = [
    5, 4, 6,
    1, 8, 7,
    3, 2, 0
]

# Goal state
goal = [
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
]


# Solve the puzzle
solution = ids(start, goal)


# Display solution
print("\nSolution Found!")
print("Number of steps:", len(solution) - 1)
print()

for i, state in enumerate(solution):
    print("Step", i)
    print_puzzle(state)
```
